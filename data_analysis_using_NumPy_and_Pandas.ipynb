```python
 ## Pandas_Series ##


import pandas as pd
index = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
marks = pd.Series([95, 92, 89, 85, 80], index=index)
print(marks)
print("\n")

print("1st Rank Mark:", marks.iloc[0])
print("\n")

print("Top 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
print("\n")
print("3rd Rank Mark:", marks.iloc[2])
print("\n")

print("the marks that are greater than 90:")
print(marks[marks > 90])
print("\n")

# 1. Modify the mark of the student with the 1st rank to 100
marks.loc['Rank1'] = 100
print("Modified Series:")
print(marks)
print("\n")

# 2. Remove the entry corresponding to the last rank from the Series
marks = marks.drop('Rank5')
print("Series after removing Rank5:")
print(marks)
print("\n")

# 3. Compute the CGPA by dividing each mark in the Series by 10
cgpa = marks / 10
print("Computed CGPA:")
print(cgpa)
```

```python
# 1. Creating Pandas DataFrame named transactions
import pandas as pd

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}


transactions = pd.DataFrame(data)
display(transactions)

# 2. Data Exploration
print("\n")
print("--- Head ---")
display(transactions.head())

print("\n--- Tail ---")
display(transactions.tail())

print(f"\nShape: {transactions.shape}")
print(f"Column Names: {list(transactions.columns)}")
print("\n--- Data Types & Info ---")
transactions.info()
```

```python
# Display only the 'ProductCategory' and 'Amount' columns
display(transactions[['ProductCategory', 'Amount']])
```

```python
# Retrieve the last 3 columns of the DataFrame
# (ProductCategory, Region, Amount are the last 3 columns)
last_three_cols = transactions.iloc[:, -3:]
display(last_three_cols)
```

```python
# Filter rows where 'Region' is 'North' and 'Amount' is greater than 200
filtered_df = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
display(filtered_df)
```

```python
# Find the value counts for the 'ProductCategory' column
print("Product Category Value Counts:")
print(transactions['ProductCategory'].value_counts())

# Find the unique values in the 'Region' column
print("\nUnique Regions:")
print(transactions['Region'].unique())
```

```python
# Group by 'Region' and find the mean amount for each region
mean_by_region = transactions.groupby('Region')['Amount'].mean().reset_index()
display(mean_by_region)
```

```python
# 3. Manipulating the DataFrame
# Modify the 'Amount' for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

# Add a new column 'Discount' by calculating 10% of the 'Amount'
transactions['Discount'] = transactions['Amount'] * 0.10

print("DataFrame after modification and adding Discount column:")
display(transactions)
```

```python
# Remove the row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

# Delete the 'Discount' column from the DataFrame
transactions = transactions.drop(columns=['Discount'])

print("Final DataFrame:")
display(transactions)
```
